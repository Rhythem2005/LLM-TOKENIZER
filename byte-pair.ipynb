```python
import tiktoken
tokenizer = tiktoken.get_encoding("gpt2")
```

```python
text = """The quick brown fox jumps over the lazy dog. This sentence contains uppercase and lowercase letters, numbers like 12345, punctuation marks (!, ?, ., ,), and multiple spaces.

Natural language processing models split text into tokens before processing it. Different tokenizers may produce different token sequences for the same input.

<|endoftext|>"""
integers = tokenizer.encode(text, allowed_special={"<|endoftext|>"})
print(integers)
```

```python
strings = tokenizer.decode(integers)
print(strings)
```
